```python
#QUANTAS VEZES CADA MODELO FOI UTILIZADO

import pandas as pd

df_foguetes = pd.read_csv("/content/analise-lunar/Moonlanding.csv")

display(df_foguetes["Carrier Rocket"].value_counts())
```

```python
#MÉDIA DE QUANTAS VEZES UM MESMO MODELO FOI UTILIZADO
media = df_foguetes["Carrier Rocket"].value_counts().mean()
print("{:.2f}".format(media))
```
