```python
import pandas as pd

# Ler o arquivo CSV
df = pd.read_csv("Moonlanding.csv")

# Atividade 1 - Frequência
frequencia = df["Mission Type"].value_counts()

print("Quantidade de missões por tipo:")
print(frequencia)

# Atividade 2 - Mínimo
minimo = frequencia.min()

print("\nQuantidade mínima registrada para o tipo de missão menos comum:")
print(minimo)
```
